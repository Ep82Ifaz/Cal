```python
def addie(a,b):
    sum = a+b
    return sum
```

```python
def minus(a,b):
    sub = a-b 
    return sub
```

```python
def multi(a,b):
    multiplied = a*b
    return multiplied
```

```python
def dev(a,b):
    devided = a/b
    return devided
```
